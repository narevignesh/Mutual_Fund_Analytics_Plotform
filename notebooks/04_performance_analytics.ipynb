**Formulas:** CAGR, Sharpe ($R_f=6.5\%$), Sortino, Alpha, Beta, Tracking Error, Drawdown


### 1. Financial Performance Metric Definitions
$$\text{CAGR} = (\text{Ending NAV} / \text{Start NAV})^{1/\text{Years}} - 1$$
$$\text{Sharpe} = \frac{R_p - R_f}{\sigma_p}, \quad \text{Sortino} = \frac{R_p - R_f}{\sigma_{\text{downside}}}$$
$$\text{Composite Score} = 0.30 R_{3Y} + 0.25 \text{Sharpe} + 0.20 \alpha + 0.15 \text{TER}_{\text{inv}} + 0.10 \text{MDD}_{\text{inv}}$$
